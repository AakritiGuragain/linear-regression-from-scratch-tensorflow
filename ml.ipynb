```python
import tensorflow as tf
import matplotlib.pyplot as plt

print(tf.__version__)
```

```python
a = tf.constant(3)
b= tf.constant([1.0,3.0])
c= tf.constant ([[1.0,2.0],[3.0,4.0]])

print("Scalar:", a)
print("Vector:", b)
print("Matrix:", c)
```

```python
x = tf.constant([[1,2]])
y= tf.constant ([[3],[4]])

result = tf.matmul(x,y)
print('Matrix Multipliccaction:', result.numpy())
```

```python
import numpy as np
X= np.array([2,3,4,5,6,7], dtype=float)
Y= np.array([6,7,8,9,1,2], dtype= float)

w = tf.Variable(0.0)
b = tf.Variable(0.0)

lr= 0.01

for epoch in range(1000):
  with tf.GradientTape() as tape:
    y_pred = w * X + b
    loss= tf.reduce_mean(tf.square(Y-y_pred))

    gradients= tape.gradient(loss,[w,b])
    w.assign_sub(lr*gradients[0])
    b.assign_sub(lr*gradients[1])

print(f"Trained W: {w.numpy()}, b: {b.numpy()}")
```

```python
plt.figure(figsize=(6,4))

# Scatter plot of original data
plt.scatter(X, Y)

# Regression line
plt.plot(X, w.numpy() * X + b.numpy())

plt.xlabel("X")
plt.ylabel("Y / Prediction")
plt.title("Linear Regression Fit")
plt.grid(True)
plt.show()
```
